
# Employee Turnover Prediction at Salifort Motors

## Business Problem
Salifort Motors is experiencing a high rate of employee turnover, leading to increased hiring costs and organizational disruption.

## Objective
- Predict whether an employee will leave the company  
- Identify key drivers of turnover  
- Provide actionable, ethical recommendations to leadership



## Data Loading & Overview
We begin by loading the HR dataset and inspecting its structure.


```

import pandas as pd

df = pd.read_csv("HR_capstone_dataset.csv")
df.head()
```

```

df.info()
```


## Exploratory Data Analysis (EDA)


```

import matplotlib.pyplot as plt

df['left'].value_counts(normalize=True).plot(kind='bar')
plt.title("Employee Turnover Rate")
plt.xlabel("Employee Left (0 = Stayed, 1 = Left)")
plt.ylabel("Proportion")
plt.show()
```

```

df.groupby('left')['satisfaction_level'].mean().plot(kind='bar')
plt.title("Average Satisfaction by Turnover Status")
plt.xlabel("Employee Left")
plt.ylabel("Average Satisfaction")
plt.show()
```

```

df.groupby('left')['average_montly_hours'].mean().plot(kind='bar')
plt.title("Average Monthly Hours by Turnover Status")
plt.xlabel("Employee Left")
plt.ylabel("Average Monthly Hours")
plt.show()
```


## Feature Engineering


```

df_encoded = pd.get_dummies(df, columns=['Department', 'salary'], drop_first=True)

X = df_encoded.drop('left', axis=1)
y = df_encoded['left']
```


## Train-Test Split


```

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```


## Logistic Regression Model
Chosen for interpretability and transparency.


```

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score

log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)

y_pred = log_model.predict(X_test)
y_prob = log_model.predict_proba(X_test)[:,1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```


## Random Forest Model (Validation)


```

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
rf_prob = rf.predict_proba(X_test)[:,1]

print(classification_report(y_test, rf_pred))
print("ROC-AUC:", roc_auc_score(y_test, rf_prob))
```


## Business Insights & Recommendations

- Low satisfaction is the strongest predictor of turnover  
- Excessive workloads increase attrition risk  
- Lack of promotions contributes significantly to employee departures  

### Recommendations
- Monitor satisfaction regularly  
- Balance workloads  
- Improve promotion transparency  

### Ethical Note
Predictions indicate risk, not certainty. Models should support employees, not penalize them.
